```python
import originpro as op
import numpy as np
import pandas as pd

import os
# Very useful, especially during development, when you are
# liable to have a few uncaught exceptions.
# Ensures that the Origin instance gets shut down properly.
# Note: only applicable to external Python.
import sys
def origin_shutdown_exception_hook(exctype, value, traceback):
    '''Ensures Origin gets shut down if an uncaught exception'''
    op.exit()
    sys.__excepthook__(exctype, value, traceback)
if op and op.oext:
    sys.excepthook = origin_shutdown_exception_hook

import  re
# Set Origin instance visibility.
# Important for only external Python.
# Should not be used with embedded Python.
if op.oext:
    op.set_show(True)


# Example of opening a project and reading data.

# We'll open the Tutorial Data.opju project that ships with Origin.
src_opju = r"C:\usrspace\mywork\linkrule.opju"   # <— 用绝对路径；注意路径和文件名是否正确
print("File exists? ", os.path.exists(src_opju))
ok = op.open(file=src_opju)
print("Project opened? ", ok)
import draw.pymatlab2.origin_function.csv2origin as csv2origin

import draw.pymatlab2.origin_function.fillgap as fillgap
from config import DATA_DIR,INPUT_DIR
from pathlib import Path
```

```python


BOOKname = "Book1"
SRC_DIR = Path(r"C:\usrspace\mywork\data\fixed_coordination_modify\Azangleslewrate")

def natural_key(p: Path):
    # 自然排序：QF_01_17_QF_02_15_... 会按数字顺序排
    return [int(t) if t.isdigit() else t.lower() for t in re.split(r'(\d+)', p.name)]

def sanitize_sheet_name(name: str, maxlen: int = 64) -> str:
    # 仅保留字母/数字/下划线，其他转为下划线，并限制长度
    name = re.sub(r'[^A-Za-z0-9_]', '_', name)
    return name[:maxlen]

csv_files = sorted(SRC_DIR.glob("*.csv"), key=natural_key)

for csv_path in csv_files:
    sheet = sanitize_sheet_name(csv_path.stem)
    try:
        csv2origin.csv_to_origin(csv_path, book=BOOKname, sheet=sheet, clear=True)
        print(f"OK  -> {csv_path.name}  ==>  {BOOKname}:{sheet}")
    except Exception as e:
        print(f"FAIL -> {csv_path.name} : {e}")
```

这里，我们是绘制方位角加速度

```python
import draw.pymatlab2.origin_function.write2origin as write2origin
import re
SRC_BOOK = 'Book1'   # 角速度
DST_BOOK = 'Book4'   # 角加速度输出



def ensure_book(name: str):
    bk = op.find_book('w', name)
    if bk is None:
        raise RuntimeError(f"找不到工作簿 '{SRC_BOOK}'")

    return bk
def accel_from_rate(rate: np.ndarray, dt: float = 1.0) -> np.ndarray:
    acc = np.empty_like(rate, dtype=float)
    acc[0] = 0.0                    # 若想用 NaN：np.nan
    acc[1:] = np.diff(rate) / dt
    return acc
# 1) 找到源/目标工作簿
bk_src = ensure_book(SRC_BOOK)
bk_dst = ensure_book(DST_BOOK)

#
# bk = op.find_book('w', SRC_BOOK)
# if bk is None:
#     raise RuntimeError(f"找不到工作簿 '{SRC_BOOK}'")

results = []




for wks in bk_src:
    df = wks.to_df()  # 读取为 DataFrame（要求列里有 time 和 AngleRate）
    # 兜底：若列名不同，默认取第1列为 time、第2列为角速度
    tcol = 'time' if 'time' in df.columns else df.columns[0]
    ycol = 'AngleRate' if 'AngleRate' in df.columns else df.columns[1]

    t = pd.to_numeric(df[tcol], errors='coerce').to_numpy()
    w = pd.to_numeric(df[ycol], errors='coerce').to_numpy()

    # Δt（若你的 time 就是步长秒数，通常为1；这里仍做一下鲁棒处理）
    if len(t) >= 2:
        dt_series = np.diff(t)
        dt = float(dt_series[~np.isnan(dt_series)][0]) if np.allclose(dt_series, dt_series[0], equal_nan=True) else 1.0
    else:
        dt = 1.0

    a = accel_from_rate(w, dt=dt)
    out = pd.DataFrame({
        'time': t,
        'AngleAccel': a
    })

    # 3) 写入目标工作簿同名 Sheet
    try:
        wks_out = bk_dst.add_sheet(wks.name)   # 新建同名表
    except Exception:
        # 已存在则清空后复用
        wks_out = bk_dst.sheet(wks.name)
        wks_out.clear()

    wks_out.from_df(out)

    # 设定列角色/单位/长名称（用 LabTalk，确保作用到当前表）
    wks_out.activate()
    op.lt_exec('col(1).type=4;')  # X
    op.lt_exec('col(1).lname$="Time"; col(1).units$="s";')
    op.lt_exec('col(2).type=1;')  # Y
    op.lt_exec('col(2).name$="AngleAccel";')
    op.lt_exec('col(2).lname$="Azimuth angle acceleration";')
    op.lt_exec('col(2).units$="deg/s^2";')

print('Done: 所有角速度表已计算并写入 Book4。')
```
